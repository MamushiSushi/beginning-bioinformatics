# **<font color=blue>P4. </font>** Data types: use ```type()``` to find a variable's "type"
####**TASK P4.1:** Ask Gemini, "*Give me examples of each data type to see how they look as outputs.*"

### Integer (`int`)
Used for whole numbers (positive, negative, or zero).

````python
my_integer = 100
print(f"Value: {my_integer}, Type: {type(my_integer)}")
````

### Float (`float`)
Used for numbers with a decimal point.

````python
my_float = 3.14159
print(f"Value: {my_float}, Type: {type(my_float)}")
````

### String (`str`)
Used for sequences of characters (text), enclosed in single or double quotes.

````python
my_string = "Hello, Python!"
print(f"Value: '{my_string}', Type: {type(my_string)}")
````

### Boolean (`bool`)
Used for logical values: `True` or `False`.

````python
my_boolean = True
print(f"Value: {my_boolean}, Type: {type(my_boolean)}")
````

### List (`list`)
An ordered, mutable collection of items. Items are enclosed in square brackets.

````python
my_list = [1, "apple", 3.0, False]
print(f"Value: {my_list}, Type: {type(my_list)}")
````

### Tuple (`tuple`)
An ordered, immutable collection of items. Items are enclosed in parentheses.

````python
my_tuple = (1, "apple", 3.0, False)
print(f"Value: {my_tuple}, Type: {type(my_tuple)}")
````

### Dictionary (`dict`)
An unordered collection of key-value pairs. Enclosed in curly braces.

````python
my_dictionary = {"name": "Alice", "age": 30}
print(f"Value: {my_dictionary}, Type: {type(my_dictionary)}")
````

### Set (`set`)
An unordered collection of unique items. Enclosed in curly braces, but without key-value pairs.

````python
my_set = {1, 2, 3, 3, 4}
print(f"Value: {my_set}, Type: {type(my_set)}")
````

#<font color=blue>**P23.**</font> SeqUtils.gc_fraction( ) – a function to compute GC content

````python
from Bio.SeqIO import parse
from Bio.SeqUtils import gc_fraction

highest_GC = 0
highest_ID = ""

for seq_record in parse("rosalind_gc.txt", "fasta"):
    GC = gc_fraction(seq_record.seq) * 100

    if GC > highest_GC:
        highest_GC = GC
        highest_ID = seq_record.id

print(highest_ID)
print(highest_GC)
````
